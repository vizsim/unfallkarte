```python
import subprocess



input_geojson_gz = "accidents_germany_2017-2023_oid_grouped.geojson.gz"


# First Tippecanoe command: zoom levels 6–8
cmd1 = [
    "tippecanoe",
    "-o", "clusters_6_8_may25_group.pmtiles",
    "--minimum-zoom=6",
    "--maximum-zoom=8",
    #"--base-zoom=8",
    "--drop-rate=0",
    "--cluster-distance=25",
    "--accumulate-attribute=UKATEGORIE__1:sum",
    "--accumulate-attribute=UKATEGORIE__2:sum",
    "--accumulate-attribute=UKATEGORIE__3:sum",
    "--no-feature-limit",
    "--no-tile-size-limit",
    "--maximum-tile-bytes=1000000",
    "--force",
    "-l", "clusters_6_8",
    input_geojson_gz
]

# Second Tippecanoe command: zoom levels 9–11
cmd2 = [
    "tippecanoe",
    "-o", "clusters_9_11_may25_group.pmtiles",
    "--minimum-zoom=9",
    "--maximum-zoom=11",
    "--base-zoom=11",
    "--drop-rate=0",
    "--cluster-distance=10",
    "--accumulate-attribute=UKATEGORIE__1:sum",
    "--accumulate-attribute=UKATEGORIE__2:sum",
    "--accumulate-attribute=UKATEGORIE__3:sum",
    "--no-feature-limit",
    "--no-tile-size-limit",
    "--maximum-tile-bytes=1000000",
    "--force",
    "-l", "clusters_9_11",
    input_geojson_gz
]

# tile-join command to combine them
cmd3 = [
    "tile-join",
    "-o", "combined_may25_group.pmtiles",
    "clusters_6_8_may25_group.pmtiles",
    "--force",
    "clusters_9_11_may25_group.pmtiles"
]

# Run commands
for i, cmd in enumerate([cmd1, cmd2, cmd3], 1):
    print(f"Running command {i}...")
    subprocess.run(cmd, check=True)
```
