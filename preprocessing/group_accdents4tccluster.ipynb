```python
import geopandas as gpd

import pandas as pd
```

```python
acc_gdf = gpd.read_parquet('accidents_germany_2017-2023_oid.parquet')
```

```python
# add for tc clusterting and pie charts later

acc_gdf['UKATEGORIE__1'] = (acc_gdf['UKATEGORIE'] == 1).astype(int)
acc_gdf['UKATEGORIE__2'] = (acc_gdf['UKATEGORIE'] == 2).astype(int)
acc_gdf['UKATEGORIE__3'] = (acc_gdf['UKATEGORIE'] == 3).astype(int)
```

```python
## this takes 1min

import gzip
import shutil

# Unkomprimiert speichern
acc_gdf.to_file("accidents_germany_2017-2023_oid_grouped.geojson", driver="GeoJSON")

# Dann gzip-komprimieren
with open("accidents_germany_2017-2023_oid_grouped.geojson", 'rb') as f_in:
    with gzip.open("accidents_germany_2017-2023_oid_grouped.geojson.gz", 'wb') as f_out:
        shutil.copyfileobj(f_in, f_out)
```
