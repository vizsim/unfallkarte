```python
import subprocess

input_fgb = "germany_osm_schools-25-05-09.fgb"

cmd = [
    "tippecanoe",
    "-o", "germany_osm_schools_full_25-05-09.pmtiles",
    "--minimum-zoom=11",
    "--maximum-zoom=15",
    "--drop-rate=0",
    "--drop-densest-as-needed",
    "--no-feature-limit",
    "--no-tile-size-limit",
    "--maximum-tile-bytes=1000000",
    "--force",
    "-l", "germany_osm_schools",
    input_fgb
]

print("Running tippecanoe for schools...")
subprocess.run(cmd, check=True)
```
